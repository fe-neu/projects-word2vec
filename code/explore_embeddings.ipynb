# Exploring Word2Vec Embeddings

This notebook provides a small, interactive playground to inspect and analyze
trained Word2Vec embeddings produced by this project.

The notebook intentionally contains **no training logic** and only consumes
the existing Python abstractions.

## Setup

Adjust the run name to match a completed training run.

```python
from pathlib import Path
from embedding_retriever import EmbeddingRetriever

RUN_DIR = Path("runs")
RUN_NAME = "run_XXXXX"  # <-- change this

retriever = EmbeddingRetriever(
    run_dir=RUN_DIR,
    run_name=RUN_NAME
)
```

## Nearest Neighbors

Compare cosine similarity and Euclidean distance.

```python
retriever.get_neighbours("king", method="cosine_similarity")[:10]
```

```python
retriever.get_neighbours("king", method="euclidean_distance")[:10]
```

## Vector Arithmetic

Classic Word2Vec example: `king - man + woman`

```python
v = (
    retriever.get_embedding("king")
    - retriever.get_embedding("man")
    + retriever.get_embedding("woman")
)

retriever.get_similar_words(v, method="cosine_similarity")[:10]
```

## Notes

- Cosine similarity tends to emphasize semantic direction
- Euclidean distance is more sensitive to vector magnitude
- Results depend heavily on corpus size and training duration
